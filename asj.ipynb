```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\syedf\OneDrive\Documents\Personal\Syed1.png")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

height, width = img.shape[:2]

plt.imshow(img_rgb)
plt.title("Image with Text")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("Original RGB Image")
plt.axis("off")
plt.show()
```

```python
img_hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
plt.figure(figsize=(8, 6))
plt.imshow(img_hsv)
plt.title("HSV Image")
plt.axis("off")
plt.show()
```

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.figure(figsize=(8, 6))
plt.imshow(gray_img, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
img_ycrcb = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
plt.figure(figsize=(8, 6))
plt.imshow(img_ycrcb)
plt.title("YCrCb Image")
plt.axis("off")
plt.show()
```

```python
img_rgb = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2RGB)
plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("HSV to RGB Image")
plt.axis("off")
plt.show()
```

```python
x = (w - 600) // 2
y = (h - 600) // 2
cv2.rectangle(img, (x, y), (x + 600 - 1, y + 600 - 1),
             (255, 255, 255), thickness=-1)
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Image with Center White Block")
plt.axis("off")
plt.show()
cv2.imwrite("output.png", img)
print("Image saved as output.png")
```

```python
img = cv2.imread(r"C:\Users\syedf\OneDrive\Documents\Personal\Syed1.png")
resized = cv2.resize(img, (img.shape[1] // 2, img.shape[0] // 2))
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Image Resized (Half Size)")
plt.axis("off")
plt.show()
```
